```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Mese": ["Gen", "Feb", "Mar", "Apr", "Mag"],
    "Vendite": [120, 150, 170, 160, 200]
}

df = pd.DataFrame(data)

df
```

```python
df.plot(x="Mese", y="Vendite", kind="bar")

plt.title("Vendite mensili")

plt.show()
```

# Sales Data Analysis

This project analyzes a small dataset of monthly sales using Python.

Tools used:
- Python
- Pandas
- Matplotlib

Goal:
Visualize monthly sales and understand trends in the data.

```python
df.plot(x="Mese", y="Vendite", kind="bar", legend=False)

plt.title("Andamento vendite mensili")
plt.xlabel("Mese")
plt.ylabel("Vendite")

plt.show()
```
